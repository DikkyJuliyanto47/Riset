```python
from madrI_config import (
    FLOOD_SHP_PATH,
    EVAC_GEOJSON_PATH,
    USER_SHP_PATH,
    ELEVATION_TIF_PATH,
    NUM_AGENTS,
    NUM_EPISODES,
)
from madrI_utils import (
    load_flood_polygons,
    load_evac_candidates_geojson,
    sample_users_from_shp,
)
from madrI_trainer import train_madrl_one_shot
```

```python
print("Memuat data...")
flood_gdf = load_flood_polygons(FLOOD_SHP_PATH)
evac_candidates = load_evac_candidates_geojson(EVAC_GEOJSON_PATH)
```

```python
# Precompute elevasi kandidat (opsional)
from madrI_utils import get_elevation_at_point

for c in evac_candidates:
    lat, lon = c["coord"]
    c["elev"] = get_elevation_at_point(ELEVATION_TIF_PATH, lat, lon)
```

```python
# sample user pool (misal 500 users)
print("Sampling user pool...")
user_pool = sample_users_from_shp(USER_SHP_PATH, n=500)
```

```python
# Run training (akan verbose)
actor, critic, history = train_madrl_one_shot(
    evac_candidates,
    flood_gdf,
    ELEVATION_TIF_PATH,
    user_pool,
    num_agents=NUM_AGENTS,
    num_episodes=NUM_EPISODES,
)
```

```python
# Simpan model actor
import torch

torch.save(actor.state_dict(), "madrI_actor_final.pt")
torch.save(critic.state_dict(), "madrI_critic_final.pt")
print("Selesai. Model tersimpan: madrI_actor_final.pt, madrI_critic_final.pt")
```
